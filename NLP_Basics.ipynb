Metinlerde bulunan fazla boşlukları ortadan kaldıralım.

```python
text="Hello    World!  2035  "
print(text)
print(text.split())
cleaned_text1=" ".join(text.split()) #Bir tane boşluk bırakmazsan hepsi bitişik yazar.Bu hoş değil.
print(f"first text: {text} \n cleaned_text1: {cleaned_text1}")
```

Büyük küçük harf çevirisi yapalım.

```python
text2="HEllo, World! 2035"
cleaned_text2=text2.lower()
```

```python
print(f"first tex2:{text2} \n cleaned text2:{cleaned_text2}")
```

Noktalama işaretlerini kaldırmayı öğrenelim.

```python
import string
text="hello,, world! 2035"
cleaned_text3=text.translate(str.maketrans("","",string.punctuation)) #Ordaki içi boş tırnaklar hiçbir şeyi hiçbir şeyle değiştirme demek.
print(f"first text3:{text} \n cleaned text3:{cleaned_text3}")
```

Özel karakterleri kaldıralım  % @ / *

```python
import re  
text = "Hello, World! 2035% "  
cleaned_text4 = re.sub(r"[^A-Za-z0-9\s]", "", text)  
print(f"text: {text} \n cleaned_text4: {cleaned_text4}")
```

Yazım hatalarını düzeltelim.

```python
from textblob import TextBlob  #Metin analizlerinde kullanılan bir kütüphane,çok faydalıdır.
```

```python
text="Hellio Wirld ! 2028"
cleaned_text5=TextBlob(text).correct() #Correct yazım hatalarını düzeltmeye yarar.
print("Düzeltilmiş metin: ",cleaned_text5)
```

html etiketlerini kaldırmayı öğrenelim.

```python
from bs4 import BeautifulSoup
html_text = "<p>merhaba <b>güzel</b> insan. <a href='#'>Bağlantı</a> içerir.</p>"
# HTML etiketlerini kaldır
clean_text6 = BeautifulSoup(html_text, "html.parser").get_text()
print(clean_text6)
```
