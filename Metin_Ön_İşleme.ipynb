Tokenizasyon işlemi gerçekleştirelim.
Bu işlemi yaparken fonksiyon noktalama işaretlerini ve boşlukları ayrı birer token olarak ayırır.
Dolayısıyla ilk önce veri temizleme yapıp sonradan tokenziayon yapmak daha doğrudur.

```python
import nltk
text="Ben veri bilimi öğreniyorum,kendimi geliştiriyorum!."
word_tokens=nltk.word_tokenize(text)
sent_tokens=nltk.sent_tokenize(text)
print("kelime tokenları: ",word_tokens)
print("cümle tokenları: ",sent_tokens)
```

Stemming ve lemmatizastion işlemlerini öğrenelim.

```python
import nltk
# WordNet veritabanını indir (Lemmatization işlemi için gerekli)
nltk.download("wordnet")
# Stemming için PorterStemmer'ı içe aktar
from nltk.stem import PorterStemmer
# Porter Stemmer nesnesini oluştur
stemmer = PorterStemmer()
# Örnek kelimeler
words = ["running", "runner", "ran", "runs", "better", "go", "went"]
# Kelimelerin köklerini bul (Stemming)
stems = [stemmer.stem(w) for w in words]
print(f"Stems: {stems}")
# %% Lemmatization işlemi
# WordNetLemmatizer'ı içe aktar
from nltk.stem import WordNetLemmatizer
# Lemmatizer nesnesini oluştur
lemmatizer = WordNetLemmatizer()
# Aynı kelime listesi ile lemmatization işlemi
lemmas = [lemmatizer.lemmatize(w,pos="v") for w in words] #pos="v" komutu Verb e göre yapmasını söyler
print(f"Lemmas: {lemmas}")
```

Stop Words

```python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")
```

ingilizce kelimelerin stopwords analizini yapalım.

```python
stop_words_eng=set(stopwords.words("english"))
print(stop_word_eng)
```

```python
#Örnek metin
text="there are some examples of handling stopwords from some text"
text_list=text.split()
text_list
```

```python
# İngilizce stop words filtresi
filtered_words = [word for word in text_list if word.lower() not in stop_words_eng]
print(f"filtered_words: {filtered_words}")
```

```python
# Türkçe stop words analizi (nltk)
stop_words_tr = set(stopwords.words("turkish"))
print("Türkçe stop kelimeleri: ",stop_words_tr)
# Örnek Türkçe metin
metin = "Merhaba arkadaşlar çok güzel bir ders işliyoruz. Bu ders faydalı mı?"
metin_list = metin.split()

filtered_words_tr = [word for word in metin_list if word.lower() not in stop_words_tr]
filtered_words_tr
```

```python
# Kütüphanesiz stop words çıkarımı
# Stop word listesi oluştur
tr_stopwords = {"için", "bu", "ile", "mi", "özel"}
# Örnek Türkçe metin
metin = "Bu bir denemedir. Amacımız bu metinde bulunan özel karakterleri elemek mi acaba?"
metin_list = metin.split()
filtered_words = [word for word in metin_list if word.lower() not in tr_stopwords]
#eğer stop wordleri filtrelemek istiyorsan not in deki in i kaldır
filtered_stop_words = [word for word in metin_list if word.lower()  in tr_stopwords]

print(filtered_words)
print(filtered_stop_words)
```
